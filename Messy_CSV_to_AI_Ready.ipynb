```python
# Messy CSV → Clean AI-Ready Data
# Portfolio Project by Sudhakar

import pandas as pd
import numpy as np
from IPython.display import display

# Better table display
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.max_colwidth', 100)

print("=== Step 1: Creating a Messy Sample Dataset ===")

# Creating intentional messy data
data = {
    'Customer Name': ['  John Doe ', 'jane smith', 'ALICE WONDER', 'Bob  ', 'John Doe', np.nan, 'Charlie'],
    'Email': ['john@email.com', 'JANE@EMAIL.COM', 'alice@email', 'bob@email.com', 'john@email.com', 'test@test.com', None],
    'Age': ['25', 'thirty', '28', '35', '25', '40', '22'],
    'Purchase Amount': ['$1,200.50', '850', '2,300.00', 'unknown', '1200.5', '950.75', '1,100'],
    'Join Date': ['2023-01-15', '15/02/2023', '2023/03/10', 'April 5 2023', '2023-01-15', '2023-06-20', '01-07-2023'],
    'Status': ['Active', 'active', 'INACTIVE', 'Active', 'Active', 'pending', 'Active']
}

df_messy = pd.DataFrame(data)

print("\nMessy Data (Before Cleaning):")
display(df_messy)

print("\n=== Step 2: Cleaning the Data ===")

df = df_messy.copy()

# 1. Clean Customer Name
df['Customer Name'] = df['Customer Name'].astype(str).str.strip().str.title()
df['Customer Name'] = df['Customer Name'].replace('Nan', 'Unknown')
df['Customer Name'] = df['Customer Name'].fillna('Unknown')

# 2. Clean Email
df['Email'] = df['Email'].astype(str).str.lower().str.strip()
df['Email'] = df['Email'].replace('alice@email', 'alice@email.com')
df['Email'] = df['Email'].replace('none', np.nan)

# 3. Clean Age
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Age'] = df['Age'].fillna(df['Age'].median())

# 4. Clean Purchase Amount
df['Purchase Amount'] = df['Purchase Amount'].astype(str).replace('[\$,]', '', regex=True)
df['Purchase Amount'] = pd.to_numeric(df['Purchase Amount'], errors='coerce')
df['Purchase Amount'] = df['Purchase Amount'].fillna(df['Purchase Amount'].median())

# 5. Clean Status
df['Status'] = df['Status'].astype(str).str.strip().str.title()

# 6. Remove exact duplicates
df = df.drop_duplicates()

print("\nCleaned Data (After Cleaning):")
display(df)

print("\n=== Step 3: Making it AI-Ready ===")

# Create AI-ready text column
df['AI_Text'] = (
    "Customer: " + df['Customer Name'] +
    " | Email: " + df['Email'].fillna('Not provided') +
    " | Age: " + df['Age'].astype(int).astype(str) +
    " | Purchase: $" + df['Purchase Amount'].round(2).astype(str) +
    " | Status: " + df['Status']
)

print("\nAI-Ready Version:")
display(df[['Customer Name', 'AI_Text']])

print("\n=== Cleaning Summary ===")
print(f"Original rows: {len(df_messy)}")
print(f"Cleaned rows: {len(df)}")
print("Missing values handled: Yes")
print("Duplicates removed: Yes")
print("AI-ready text column created: Yes")
```
