```python
# Messy Cafe Sales Data → Clean AI-Ready Data
# Portfolio Project by Sudhakar

import pandas as pd
import numpy as np
from IPython.display import display

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.max_colwidth', 100)

print("=== Step 1: Loading Real Messy Cafe Sales Data ===")

df_messy = pd.read_csv('dirty_cafe_sales.csv')

print("\nShape of original data:", df_messy.shape)
print("\nColumn names:", df_messy.columns.tolist())
print("\nMessy Data (First 8 rows):")
display(df_messy.head(8))

print("\nMissing values:")
print(df_messy.isnull().sum())

print("\n=== Step 2: Cleaning the Data ===")

df = df_messy.copy()

# Clean Item
if 'Item' in df.columns:
    df['Item'] = df['Item'].astype(str).str.strip().str.title()
    df['Item'] = df['Item'].replace(['Nan', 'None', 'nan', 'UNKNOWN', 'Unknown'], np.nan)
    df['Item'] = df['Item'].fillna('Unknown Item')

# Clean Quantity
if 'Quantity' in df.columns:
    df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
    df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())

# Clean Price Per Unit
if 'Price Per Unit' in df.columns:
    df['Price Per Unit'] = pd.to_numeric(df['Price Per Unit'], errors='coerce')
    df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Price Per Unit'].median())

# Clean Total Spent
if 'Total Spent' in df.columns:
    df['Total Spent'] = pd.to_numeric(df['Total Spent'], errors='coerce')
    df['Total Spent'] = df['Total Spent'].fillna(df['Total Spent'].median())

# Clean Payment Method
if 'Payment Method' in df.columns:
    df['Payment Method'] = df['Payment Method'].astype(str).str.strip().str.title()
    df['Payment Method'] = df['Payment Method'].replace(['Nan', 'None', 'nan', 'ERROR', 'Unknown'], 'Unknown')
    df['Payment Method'] = df['Payment Method'].fillna('Unknown')

# Clean Location
if 'Location' in df.columns:
    df['Location'] = df['Location'].astype(str).str.strip().str.title()
    df['Location'] = df['Location'].replace(['Nan', 'None', 'nan', 'UNKNOWN'], 'Unknown')
    df['Location'] = df['Location'].fillna('Unknown')

# Remove duplicates
df = df.drop_duplicates()

print("\nCleaned Data (First 8 rows):")
display(df.head(8))
print("\nShape after cleaning:", df.shape)

print("\n=== Step 3: Creating AI-Ready Version ===")

df['AI_Text'] = (
    "Item: " + df['Item'].astype(str) +
    " | Qty: " + df['Quantity'].astype(int).astype(str) +
    " | Price: $" + df['Price Per Unit'].round(2).astype(str) +
    " | Total: $" + df['Total Spent'].round(2).astype(str) +
    " | Payment: " + df['Payment Method'].astype(str) +
    " | Location: " + df['Location'].astype(str)
)

print("\nAI-Ready Sample:")
display(df[['Item', 'AI_Text']].head(8))

print("\n=== Cleaning Summary ===")
print(f"Original rows: {len(df_messy)}")
print(f"Cleaned rows: {len(df)}")
print("Missing values handled: Yes")
print("Duplicates removed: Yes")
print("AI-ready text column created: Yes")
```
